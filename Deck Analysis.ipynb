# Deck Analysis

```python
import pandas as pd

card_df = pd.read_csv("EN_Card_Data.csv")
deck_df = pd.read_csv(
    "crustle_deck.csv",
    header=None,
).squeeze("columns")
```

```python
card_df["Stage (Pokémon)/Type (Energy and Trainer)"].value_counts()
```

```python
card_df["Stage (Pokémon)/Type (Energy and Trainer)"].value_counts()
```

```python
print("count:", len(deck_df))
deck_df.value_counts()
```

```python
# Count each card ID in the deck
counts = (
    deck_df.value_counts()
    .rename_axis("Card ID")
    .reset_index(name="Count")
)

# Join with the card database
cards_in_deck = counts.merge(
    card_df[
        [
            "Card ID",
            "Card Name",
            "Category",
            "Stage (Pokémon)/Type (Energy and Trainer)",
            "Rule",
            "HP",
            "Type",
            "Weakness",
            "Resistance (Type)",
            "Retreat",
            "Move Name",
            "Cost",
            "Damage",
        ]
    ],
    on="Card ID",
    how="left"
)

# Reorder columns
cards_in_deck = cards_in_deck[
    [
        "Card ID",
        "Card Name",
        "Category",
        "Stage (Pokémon)/Type (Energy and Trainer)",
        "Rule",
        "HP",
        "Type",
        "Weakness",
        "Resistance (Type)",
        "Retreat",
        "Move Name",
        "Cost",
        "Damage",
        "Count",
    ]
]

cards_in_deck
```

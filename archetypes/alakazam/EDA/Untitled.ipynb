# Parquet Imitation Learning

```python
import pandas as pd

df = pd.read_parquet("data/policy_decisions.parquet", engine="pyarrow")
```

```python
# df.head()
```

```python
episode = df[df["episode_id"] == 87783888]
```

```python
import numpy as np
from pprint import pprint

def to_python(obj):
    """Recursively convert NumPy arrays to native Python objects."""
    if isinstance(obj, np.ndarray):
        return [to_python(x) for x in obj.tolist()]
    elif isinstance(obj, dict):
        return {k: to_python(v) for k, v in obj.items()}
    elif isinstance(obj, list):
        return [to_python(x) for x in obj]
    elif isinstance(obj, tuple):
        return tuple(to_python(x) for x in obj)
    else:
        return obj

for row in episode.itertuples(index=False):
    print("=" * 80)
    print(f"Frame: {row.frame_index}")
    print(f"Player: {row.player_name}")
    print()

    print("*" * 10)
    print("State:")
    pprint(to_python(row.state), sort_dicts=False, width=120)
    print("*" * 10)

    print("Selection:")
    pprint(to_python(row.selection), sort_dicts=False, width=120)
    print("*" * 10)

    print("Options:")
    pprint(to_python(row.options), sort_dicts=False, width=120)
    print("*" * 10)

    print("Target Action:")
    pprint(to_python(row.target_action), width=120)
    print()
```
